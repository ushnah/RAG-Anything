```python
from pathlib import Path
import sys
import pandas as pd
from dotenv import load_dotenv

ROOT = Path.cwd()
if not (ROOT / "arabic_poc").is_dir():
    ROOT = ROOT.parent
assert (ROOT / "arabic_poc").is_dir(), "Open the notebook from the repo or arabic_poc folder"

sys.path.insert(0, str(ROOT))
load_dotenv(ROOT / ".env")

from arabic_poc.person_aliases import MATCHER
from arabic_poc.query_translation import query_variants

queries = [
    "Show me a picture of MBS",
    "pribce mbs",
    "عبد الرحمن بن عبد العزيز السديس",
    "Show me photos of King Faisal",
    "Show me the green dome",
    "salman",
]

pd.DataFrame([
    {
        "Original query": q,
        "Alias-only rewrite": MATCHER.rewrite(q, "en"),
        "Full pipeline variants": query_variants(q),
    }
    for q in queries
])
```
